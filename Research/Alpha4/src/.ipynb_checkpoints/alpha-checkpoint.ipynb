```python
# 1. 初始化执行器
alpha_pool=[]
executor = AlphaExecutor(data_dir='../data/20251231',alpha_dir='../alphas/20251231')
executor.load_all_data()
executor.load_all_simres()
```

```python
for btresult in executor.alpha_pool:
    executor.simres_cut(btresult,'20150101','20260101',index=None)
```

```python
expr='ts_ols(ts_skewness(ret1,5),ret1,5)[0]'
alpha01 = executor.evaluate(f'cs_booksize(cs_rank(at_mask({expr},ts_fill(csi_500_weight)>0))-0.5)')
btresult=executor.backtest(alpha01)
executor.simres_cut(btresult,'20200101','20260101')
corr_df=executor.corr(btresult)
alpha_pool.append(btresult)
```

```python
expr='ts_ols(ts_kurtosis(ret1,5),ret1,5)[0]'
alpha01 = executor.evaluate(f'cs_booksize(cs_rank(at_mask({expr},ts_fill(csi_500_weight)>0))-0.5)')
btresult=executor.backtest(alpha01)
executor.simres_cut(btresult,'20200101','20260101')
corr_df=executor.corr(btresult)
alpha_pool.append(btresult)
```

```python

# 2. 设置绘图风格
sns.set_theme(style="white")

# 3. 绘制热力图
plt.figure(figsize=(10, 8))
sns.heatmap(np.corrcoef([item['net_ret'] for item in alpha_pool]), 
            annot=True,       # 显示具体数值
            fmt=".2f",        # 保留两位小数
            cmap='coolwarm',  # 冷暖色调，蓝色负相关，红色正相关
            center=0,         # 设置中心点为 0
            linewidths=.5)    # 格子间距

plt.title('Factor Correlation Matrix')
plt.show()
```
